```python
import os
import pickle

import numpy as np
import pandas as pd
import xarray as xr

from glob import glob
from pathlib import Path
from tqdm import tqdm
```

# Paths

```python
# Set paths for main directories
# Main path
path_main = Path(os.path.abspath('')).parent.parent.parent
print(path_main)
# Working path
path_work = Path(os.path.abspath(''))
print(path_work)
# Files path
path_files = os.path.join(path_work,'files')
```

# Leg 1 standard run

```python
leg = 1
run_name = 'standard'
# Number of drifters per cluster
npol = 5
# Time range
t1 = pd.Timestamp(2023,4,28,2)
t2 = t1 + pd.Timedelta('5 days')
time_res = '1h'
run_times = pd.date_range(t1,t2,freq=time_res)
# Limit to a certain domain?
domain_limits = True

# Create configuration dictionary
config_dict = {
    'leg':leg,'ndrifters':npol,'run_times':run_times,'domain_limits':domain_limits
}

# Save config file
fname = f'config_file_run_{run_name}_leg{leg}.pkl'
fpath = os.path.join(path_files,fname)
with open(fpath,'wb') as f:
    pickle.dump(config_dict,f)
```
